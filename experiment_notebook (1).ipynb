# Eksperimen Analisis Data
Notebook ini memuat eksperimen awal dengan data `gabungan.csv`.

**Import Library**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
file_path = 'gabungan.csv'
df = pd.read_csv(file_path)

# Tampilkan info dasar
df.info()
df.head()
```

**Penjelasan**
* `pandas` → untuk memproses dan menganalisis data

* `matplotlib.pyplot` → untuk membuat grafik

* `seaborn` → untuk visualisasi statistik yang lebih menarik

**Periksa Missing Values**

```python
# Periksa missing values
df.isnull().sum()
```

**Penjelasan**
* Mengecek berapa banyak nilai kosong (null) di tiap kolom. Ini penting untuk mengetahui apakah perlu pembersihan data.



**Statistik Deskriptif**

```python
# Statistik deskriptif
df.describe()
```

**Penjelasan**
* Menampilkan statistik ringkasan (mean, std, min, max, dll.) untuk kolom numerik.

**Heatmap Korelasi**

```python
# Heatmap korelasi
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

**Penjelasan**
*  Membuat peta panas (heatmap) untuk melihat korelasi antar kolom numerik. Warna memperlihatkan kekuatan dan arah hubungan.

**Pairplot Kolom Numerik (Jika Sedikit)**

```python
# Pairplot jika kolom numerik tidak terlalu banyak
num_cols = df.select_dtypes(include='number').columns
if len(num_cols) <= 5:
    sns.pairplot(df[num_cols])
    plt.show()
```

**Penjelasan**
* Jika jumlah kolom numerik ≤ 5, buat pairplot (scatter plot matrix) untuk melihat hubungan antar kolom secara visual.

```python
# Hitung nilai unik pada kolom kategorikal
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    print(f'\nValue counts for {col}:')
    print(df[col].value_counts())
```
